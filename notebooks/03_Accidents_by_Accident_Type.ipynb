# 사고유형별 교통사고 2025 분서

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv("../data/사고유형별_교통사고-2025.csv", header=1)
df.head(10)
```

```python
df.info()
```

```python
# df.describe

# change column names to more descriptive names
df.columns = ['사고유형별(1)', '사고유형별(2)', '월별', '사고건수', '사망자수', '부상자수']

print(df.columns)
```

## 전체적인 차트 그리기

```python
plt.rcParams['font.family'] = 'AppleGothic'
plt.rcParams['axes.unicode_minus'] = False
```

```python
# 전체 데이터만 선택
df_total = df[df['월별'] == '전체'].copy()

print(df_total)
```

## 사고건수 그룹형 막대그래프

```python
# 상세 사고유형별 사고건수 시각화
plt.figure(figsize=(12, 6))

sns.barplot(data=df_total,
            x='사고유형별(2)',
            y='사고건수',
            hue='사고유형별(1)')

plt.title('사고유형별 교통사고 발생건수 2025년')
plt.xlabel('세부 사고유형')
plt.ylabel('사고건수')
plt.xticks(rotation=45)
plt.legend(title='사고유형')

plt.show()
```

## 사망자수 그룹형 막대그래프

```python
# 사고유형별 사망자수 시각화
plt.figure(figsize=(12, 6))

sns.barplot(data=df_total,
            x='사고유형별(2)',
            y='사망자수',
            hue='사고유형별(1)')

plt.title('사고유형별 교통사고 사망자수 2025년')
plt.xlabel('세부 사고유형')
plt.ylabel('사망자수')
plt.xticks(rotation=45)
plt.legend(title='사고유형')

plt.show()
```

## 부상자수 그룹형 막대그래프

```python
# 사고유형별 부상자수 시각화
plt.figure(figsize=(12, 6))

sns.barplot(data=df_total,
            x='사고유형별(2)',
            y='부상자수',
            hue='사고유형별(1)')

plt.title('사고유형별 교통사고 부상자수 2025년')
plt.xlabel('세부 사고유형')
plt.ylabel('부상자수')
plt.xticks(rotation=45)
plt.legend(title='사고유형')

plt.show()
```

## Top 10 accident types by accident count

```python
# 사고건수가 많은 상위 10개 유형
top10 = df_total.nlargest(10, '사고건수')

plt.figure(figsize=(10, 6))

sns.barplot(data=top10,
            x='사고건수',
            y='사고유형별(2)',
            hue='사고유형별(1)')

plt.title('사고유형별 교통사고 발생건수 TOP 10 2025년')
plt.xlabel('사고건수')
plt.ylabel('사고유형')

plt.show()
```

```python
# 사고 100건당 사망자수 계산
df_total['사고당 사망자수'] = (
    df_total['사망자수'] / df_total['사고건수'] * 100
)

# 사망자 비율이 높은 상위 10개 유형
top10_death = df_total.nlargest(10, '사고당 사망자수')

plt.figure(figsize=(10, 6))

sns.barplot(data=top10_death,
            x='사고당 사망자수',
            y='사고유형별(2)',
            hue='사고유형별(1)')

plt.title('사고유형별 사고 100건당 사망자수 TOP 10 2025년')
plt.xlabel('사고 100건당 사망자수')
plt.ylabel('사고유형')

plt.show()
```
