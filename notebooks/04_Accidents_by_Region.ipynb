# 시도별 교통사고 2025 KOSIS 분서

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np
```

```python
df = pd.read_csv("../data/시도별_교통사고-2025_KOSIS.csv", header=1)
df.head(10)
```

```python
print(df.info())
```

```python
print(df.describe)
```

```python
# change column names
df.columns = ['시도별', '월별', '사고건수', '사망자수', '부상자수']

print(df.columns)
```

```python
# 시도별 고유값 개수 확인
print('시도 개수:', df['시도별'].nunique())

# 시도별 고유값 출력
print(df['시도별'].unique())
```

## 전체적인 차트 그리기

```python
# 전체 데이터만 추출
df_total = df[df['월별'] == '전체'].copy()

print(df_total)
```

```python
plt.rcParams['font.family'] = 'AppleGothic'
plt.rcParams['axes.unicode_minus'] = False
```

### 1. 시도별 교통사고 발생건수

- **차트 유형:** 막대그래프 (Bar Chart)
- **분석 내용:** 지역별 교통사고 발생건수를 비교하여 사고가 많이 발생하는 지역을 확인한다.

```python
# 시도별 교통사고 발생건수 시각화
plt.figure(figsize=(12, 6))

sns.barplot(data=df_total,
            x='시도별',
            y='사고건수')

plt.title('시도별 교통사고 발생건수 2025년')
plt.xlabel('시도')
plt.ylabel('사고건수')
plt.xticks(rotation=45)

plt.show()
```

### 2. 시도별 교통사고 사망자수

- **차트 유형:** 막대그래프 (Bar Chart)
- **분석 내용:** 지역별 교통사고 사망자수를 비교하여 사망자가 많이 발생하는 지역을 확인한다.

```python
# 시도별 교통사고 사망자수 시각화
plt.figure(figsize=(12, 6))

sns.barplot(data=df_total,
            x='시도별',
            y='사망자수')

plt.title('시도별 교통사고 사망자수 2025년')
plt.xlabel('시도')
plt.ylabel('사망자수')
plt.xticks(rotation=45)

plt.show()
```

### 3. 시도별 교통사고 부상자수

- **차트 유형:** 막대그래프 (Bar Chart)
- **분석 내용:** 지역별 교통사고 부상자수를 비교하여 부상자가 많이 발생하는 지역을 확인한다.

```python
# 시도별 교통사고 부상자수 시각화
plt.figure(figsize=(12, 6))

sns.barplot(data=df_total,
            x='시도별',
            y='부상자수')

plt.title('시도별 교통사고 부상자수 2025년')
plt.xlabel('시도')
plt.ylabel('부상자수')
plt.xticks(rotation=45)

plt.show()
```

### 4. 교통사고 발생건수 상위 10개 지역

- **차트 유형:** 가로 막대그래프 (Horizontal Bar Chart)
- **분석 내용:** 교통사고 발생건수가 많은 상위 10개 지역을 순위별로 비교한다.

```python
# 교통사고 발생건수 상위 10개 지역
top10 = df_total.nlargest(10, '사고건수')

plt.figure(figsize=(10, 6))

sns.barplot(data=top10,
            x='사고건수',
            y='시도별')

plt.title('교통사고 발생건수 상위 10개 지역 2025년')
plt.xlabel('사고건수')
plt.ylabel('시도')

plt.show()
```
