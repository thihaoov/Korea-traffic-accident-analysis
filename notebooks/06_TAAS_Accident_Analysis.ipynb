# TAAS 교통사고 2025년꺼자 종합 분석

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

## I. 가해운전자 연령층별 교통사고 분석

```python
df_age_group = pd.read_excel('../TAAS_data/가해운전자_연령층별_교통사고_2025_TAAS.xlsx', sheet_name=0)

print(df_age_group)
```

```python
print(df_age_group.info())
```

```python
plt.rcParams['font.family'] = 'AppleGothic'
plt.rcParams['axes.unicode_minus'] = False
```

## 1. 가해운전자 연령층별 교통사고 발생건수, 막대그래프 (Bar Chart)

```python
# 연령층별 사고건수 시각화
plt.figure(figsize=(10, 5))

sns.barplot(data=df_age_group, x='연령층', y='사고건수') # palette='viridis'

plt.title('가해운전자 연령층별 교통사고 발생건수 2025년')
plt.xlabel('연령층')
plt.ylabel('사고건수')
plt.xticks(rotation=45)
plt.show()
```

### 막대그래프 + 꺾은선그래프 (Bar + Line Chart)

```python
# 연령층별 교통사고 현황 시각화
x = np.arange(len(df_age_group))
width = 0.35

fig, ax1 = plt.subplots(figsize=(12, 6))

# 막대그래프: 사고건수와 부상자수
ax1.bar(x - width/2, df_age_group['사고건수'],
        width, label='사고건수')

ax1.bar(x + width/2, df_age_group['부상자수'],
        width, color='orange', label='부상자수')

ax1.set_xlabel('연령층')
ax1.set_ylabel('사고건수 및 부상자수')
ax1.set_xticks(x)
ax1.set_xticklabels(df_age_group['연령층'], rotation=30)

# 꺾은선그래프: 사망자수
ax2 = ax1.twinx()
ax2.plot(x, df_age_group['사망자수'],
         color='red', marker='o', linewidth=2, label='사망자수')

ax2.set_ylabel('사망자수')

# 범례 통합
lines1, labels1 = ax1.get_legend_handles_labels()
lines2, labels2 = ax2.get_legend_handles_labels()
ax1.legend(lines1 + lines2, labels1 + labels2, loc='upper left')

plt.title('가해운전자 연령층별 교통사고 현황 2025년')
plt.tight_layout()
plt.show()
```

## II. 법규위반별 교통사고 분석

```python
df_violation = pd.read_excel('../TAAS_data/법규위반별_교통사고_2025_TAAS.xlsx', sheet_name=0)
print(df_violation)
```

## 법규위반별 교통사고 현황 막대그래프 + 꺾은선그래프 (Bar + Line Chart)

## III. 인구10만명당 교통사고 2025 TAAS 분석

```python
df_region = pd.read_excel('../TAAS_data/인구10만명당_교통사고_2025_TAAS.xlsx', sheet_name=0)
print(df_region)
```

```python
# 사고건수 기준 내림차순 정렬
df_region = df_region.sort_values(
    by='사고건수_10만명당',
    ascending=False
)
```

### 시도별 인구 10만 명당 교통사고 현황 막대그래프 + 꺾은선그래프 (Bar + Line Chart)

## IV. 연도별 교통사고 발생 추세 (1995~2025) 분석

```python
df_by_year = pd.read_excel('../TAAS_data/교통사고추세_1995-2025_TAAS.xlsx')
print(df_by_year)
```

```python
# 연도별 교통사고 추세 시각화
df_by_year = df_by_year.sort_values('연도')

# 그래프 생성
fig, ax1 = plt.subplots(figsize=(14, 6))

# 꺾은선그래프: 사고건수
ax1.plot(df_by_year['연도'], df_by_year['사고건수'],
         color='royalblue', linewidth=2, label='사고건수')

# 꺾은선그래프: 부상자수
ax1.plot(df_by_year['연도'], df_by_year['부상자수'],
         color='orange', linewidth=2, label='부상자수')

# 왼쪽 Y축 설정
ax1.set_xlabel('연도')
ax1.set_ylabel('사고건수 및 부상자수')

# 오른쪽 Y축 생성
ax2 = ax1.twinx()

# 꺾은선그래프: 사망자수
ax2.plot(df_by_year['연도'], df_by_year['사망자수'],
         color='red', linewidth=2, label='사망자수')

ax2.set_ylabel('사망자수')

# X축 연도 표시 간격 설정
ax1.set_xticks(range(1995, 2026, 2))
ax1.tick_params(axis='x', rotation=45)

# 범례 통합
lines1, labels1 = ax1.get_legend_handles_labels()
lines2, labels2 = ax2.get_legend_handles_labels()
ax1.legend(lines1 + lines2, labels1 + labels2, loc='upper right')

# 제목 및 그래프 출력
plt.title('연도별 교통사고 발생 추세 (1995~2025)')
plt.tight_layout()
plt.show()
```

```python
# 1995년과 2025년 데이터 비교
first_year = df_by_year.iloc[0]
last_year = df_by_year.iloc[-1]

# 1995년 대비 2025년 변화율 계산
for col in ['사고건수', '사망자수', '부상자수']:
    change = (last_year[col] - first_year[col]) / first_year[col] * 100
    print(f'{col} 변화율: {change:.2f}%')
```
