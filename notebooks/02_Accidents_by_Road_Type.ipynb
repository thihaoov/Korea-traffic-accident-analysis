# 도로종류별 교통사고(KOSIS) 분서

```python
import pandas as pd

df = pd.read_csv('../data/도로종류별_교통사고-2025_KOSIS.csv', header=1)
df.head(10)
```

```python
# Rename columns to more descriptive names
df.columns = ['도로종류별', '월별', '사고건수', '사망자수', '부상자수']

print(df.columns)
```

```python
print(df.info())
```

```python
# print(df.isnull().sum())
print(df.duplicated().sum())
```

## 전체적인 차트 그리기

```python
df_total = df[df['월별']=='전체']
df_total.head(10)
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

plt.rcParams['font.family'] = 'AppleGothic'
plt.rcParams['axes.unicode_minus'] = False
```

### 도로종류별 교통사고 Bar chart

```python
## 도로종류별 사고건수 그래프

plt.figure(figsize=(10, 6))

sns.barplot(data=df_total, x='도로종류별', y='사고건수')

plt.title('도로종류별 교통사고 발생건수 2025년')
plt.xlabel('도로종류')
plt.ylabel('사고건수')
plt.xticks(rotation=45)

plt.show()
```

```python
## 도로종류별 사망자수 그래프

plt.figure(figsize=(10, 6))

sns.barplot(data=df_total, x='도로종류별', y='사망자수')

plt.title('도로종류별 교통사고 발생건수 2025년')
plt.xlabel('도로종류')
plt.ylabel('사망자수')
plt.xticks(rotation=45)

plt.show()
```

```python
## 도로종류별 부상자수 그래프

plt.figure(figsize=(10, 6))

sns.barplot(data=df_total, x='도로종류별', y='부상자수')

plt.title('도로종류별 교통사고 부상자수 2025년')
plt.xlabel('도로종류')
plt.ylabel('부상자수')
plt.xticks(rotation=45)

plt.show()
```

## Line chart for monthly traffic accident by road type

```python
road_month = df[df['월별'] != '전체'].copy()

road_month = road_month[
    road_month['도로종류별'].isin(['일반국도', '지방도', '특별광역시도', '시도', '군도', '국가지원지방도', '고속국도', '기타/불명'])
]
```

```python
road_month['월순'] = road_month['월별'].str.extract(r'(\d+)\s*월').astype(int)

road_month = road_month.sort_values('월순')
```

```python
plt.figure(figsize=(12, 6))

sns.lineplot(data=road_month,
             x='월순',
             y='사고건수',
             hue='도로종류별',
             marker='o')

plt.title('도로종류별 월별 교통사고 발생건수 2025년')
plt.xlabel('월')
plt.ylabel('사고건수')
plt.xticks(range(1, 13))

plt.show()
```
